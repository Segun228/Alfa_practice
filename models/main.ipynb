```python

df2 = df1.copy()


df2['eventTime'] = pd.to_datetime(df2['eventTime'])
df2['period_start_date'] = pd.to_datetime(df2['period_start_date'])
df2['period_end_date'] = pd.to_datetime(df2['period_end_date'])

df2['default_flag'] = ((df2['period_end_date'] - df2['eventTime']).dt.days <= 120) & ((df2['period_end_date'] - df2['period_start_date']).dt.days >= 30) 
df2['default_flag'] = df2['default_flag'].astype(int)

df2['month'] = df2['eventTime'].dt.to_period('M')
```

```python
from sklearn.metrics import roc_auc_score

def gini_safe(y_true, y_score):
    mask = (~y_true.isna()) & (~y_score.isna())
    if mask.sum() == 0 or len(np.unique(y_true[mask])) == 1:
        return np.nan
    return 2 * roc_auc_score(y_true[mask], y_score[mask]) - 1
```

```python
gini_monthly = df2.groupby('month').apply(
    lambda x: pd.Series({
        'gini_final': gini_safe(x['default_flag'], x['final_model_score'])
    })
).reset_index()

print(gini_monthly)
```

```python
def bootstrap_gini_safe(df_slice, n_iter=1000, sample_frac=0.7):
    df_slice = df_slice.dropna(subset=['final_model_score', 'default_flag'])
    if len(df_slice) == 0:
        return [np.nan, np.nan]
    gini_vals = []
    n = int(len(df_slice) * sample_frac)
    for _ in range(n_iter):
        sample = df_slice.sample(n=n, replace=True)
        gini_vals.append(gini_safe(sample['default_flag'], sample['final_model_score']))
    return np.percentile(gini_vals, [2.5, 97.5])
```

```python
ci_monthly = df2.groupby('month').apply(
    lambda x: pd.Series(bootstrap_gini_safe(x, n_iter=1000, sample_frac=0.7),
                        index=['gini_ci_lower', 'gini_ci_upper'])
).reset_index()


gini_monthly = gini_monthly.merge(ci_monthly, on='month')
print(gini_monthly)
```
